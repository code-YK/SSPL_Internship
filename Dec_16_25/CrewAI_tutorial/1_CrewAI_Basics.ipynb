## CrewAI

- multiple AI agents working like a human team

- Each agent has:
    - a role (who they are)
    - a goal (what they want to achieve)
    - tools (what they can use)
    - optional memory

- And they collaborate on tasks.

## Core Concepts (VERY IMPORTANT)

### `Agent`

- An Agent is an autonomous worker.
    - Role: Researcher
    - Goal: Find accurate information
    - Backstory: You are an expert researcher

### `Task`

- A Task is a unit of work assigned to an agent.
    - Task: Research pros & cons of electric cars
    - Assigned to: Researcher agent

### `Crew`

- A Crew is a collection of agents + tasks.
- Crew = Agents + Tasks + Process

### `Process`

- Defines how agents work together.

| Process | Meaning |
|---------|---------|
| sequential | Agents work one after another |
| hierarchical | Manager agent assigns tasks |

#### CrewAI is Best for:
- Content pipelines (research → write → edit)
- Report generation
- Business workflows
- Simple automation

#### CrewAI is Not ideal for:
- Complex loops
- Fine-grained control flow
- Heavy HITL logic (LangGraph wins here)

```python
from crewai import Agent, Task, Crew, Process
```

#### Intiating LLM

```python
from crewai import LLM

llm = LLM(
    model="groq/llama-3.1-8b-instant",
    temperature=0.3
)
```

#### Creating agent

```python
researcher = Agent(
    role="Researcher",
    goal="Find accurate information on agentic AI",
    backstory="You are an AI research expert who gathers reliable facts.",
    llm=llm,
    verbose=True
)

writer = Agent(
    role="Writer",
    goal="Explain agentic AI in simple language",
    backstory="You are a technical writer who simplifies complex topics.",
    llm=llm,
    verbose=True
)
```

#### Creating Tasks

```python
research_task = Task(
    description="Research agentic AI frameworks and key ideas",
    expected_output="Bullet list of frameworks",
    agent=researcher,
    output_file="research_output.txt"
)

write_task = Task(
    description="Write a beginner-friendly explanation of agentic AI",
    expected_output="Simple explanation with examples",
    agent=writer,
    output_file="explanation_output.txt"
)
```

#### Create Crew

```python
crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, write_task],
    process=Process.sequential
)
```

```python
result = crew.kickoff()
```
