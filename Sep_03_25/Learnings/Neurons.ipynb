## NEURONS

- In deep learning, a neuron (also called a node or unit) is the basic building block of a neural network, inspired by biological neurons in the brain.

- Each neuron receives inputs, processes them, and produces an output that can be sent to other neurons.

- Mathematically, a neuron performs a 

    `weighted sum` + `bias` → `activation function` → `output`

## How a Single Neuron Works in Deep Learning

Suppose we have input features \(x_1, x_2, ..., x_n\).

---

### 1. Input & Weights  
Each input \(x_i\) is multiplied by a weight \(w_i\).  
→ Weighted inputs:  
$$
w_1x_1, \; w_2x_2, \; ..., \; w_nx_n
$$

---

### 2. Summation (Linear Combination)  
The neuron adds all weighted inputs and a **bias term \(b\)**:  

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

---

### 3. Activation Function  
To introduce non-linearity, the neuron applies an **activation function** \(f(z)\):  

$$
y = f(z)
$$

Common activation functions:
- Sigmoid: squashes values between 0 and 1.  
- ReLU: outputs 0 if negative, otherwise the input.  
- Tanh: outputs between -1 and 1.  

---

### 4. Output  
The result \(y\) is passed to the next layer (or becomes the final prediction if this is the output neuron).
