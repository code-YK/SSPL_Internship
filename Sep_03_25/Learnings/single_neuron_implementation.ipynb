# Hours Studied vs Exam Score 

- This is a simple **linear regression** example using TensorFlow.  
- build a model that predicts exam scores based on the number of hours studied.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Hours studied (features)
X = np.array([1, 2, 3, 4, 5, 6], dtype=float)

# Exam scores (labels)
y = np.array([15, 25, 35, 45, 55, 65], dtype=float)
```

#### Build the Model

- We will use a **single neuron (Dense layer with 1 unit)**.


```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])
```

#### Compile the Model

- Optimizer: Stochastic Gradient Descent (SGD)  
- Loss Function: Mean Squared Error (MSE)


```python
model.compile(optimizer='sgd', loss='mean_squared_error')
```

```python
history = model.fit(X, y, epochs=500, verbose=0)
print("Training complete!")
```

#### Plot the Loss Curve

```python
plt.plot(history.history['loss'])
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('Training Loss Curve')
plt.show()
```

```python
# Predict exam score for 7 hours of study
prediction = model.predict(np.array([[7]]))
print("Predicted score for 7 hours of study:", prediction[0][0])
```

```python
# Model predictions for range of hours
hours = np.linspace(0, 10, 100)
pred_scores = model.predict(hours)

plt.scatter(X, y, color="blue", label="Training Data")
plt.plot(hours, pred_scores, color="red", label="Model Prediction")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.legend()
plt.show()
```
