# Artificial Neural Networks

- An Artificial Neural Network (ANN) is a machine learning model inspired by the human brain.

- Just like the brain has neurons connected together, an ANN has artificial neurons (nodes) arranged in layers.

- Each neuron processes input, applies a transformation, and passes the result to the next layer.

- ANNs are used for classification, regression, image recognition, natural language processing, recommendation systems, and much more.

## Structure of an ANN:

- Input Layer → takes input features (e.g., pixel values of an image).

- Hidden Layers → perform transformations through neurons and activation functions.

- Output Layer → gives the final result (e.g., class label or probability).

## How an Artificial Neural Network (ANN) Works

Suppose we have inputs \(x_1, x_2, ..., x_n\).

---

### Step 1: Weighted Sum
Each neuron computes:

$$
z = \sum_{i=1}^n w_i x_i + b
$$

- \(w_i\) = weight (importance of feature)  
- \(b\) = bias (shifts the output)  

---

### Step 2: Activation Function
To introduce **non-linearity**, apply:

$$
a = f(z)
$$

Examples of activation functions:

- **Sigmoid:**  
  $$
  f(z) = \frac{1}{1 + e^{-z}}, \quad \text{outputs in } (0,1)
  $$
- **ReLU:**  
  $$
  f(z) = \max(0, z)
  $$
- **Tanh:** outputs between \(-1\) and \(1\).  
- **Softmax:** used in output layer for multi-class classification.  

---

### Step 3: Forward Propagation
- Input flows layer by layer (input → hidden → output).  
- Each layer transforms the data.  

---

### Step 4: Loss Function
To measure prediction error:

- **MSE (Mean Squared Error):** for regression.  
- **Cross-Entropy Loss:** for classification.  

---

### Step 5: Backpropagation
- ANN adjusts **weights & biases** to minimize loss.  
- Uses **gradient descent** (optimization algorithm).  
- Gradients are computed with **chain rule of calculus** (backpropagation).  


---

## Concept of ANN

#### a) Training Process

- Initialization → assign small random weights.

- Forward Pass → compute predictions.

- Loss Calculation → compare prediction with true label.

- Backward Pass (Backpropagation) → compute gradients.

- Update Weights → using optimizers like:

        - SGD (Stochastic Gradient Descent)

        - Adam (adaptive learning rate, very popular)

        - RMSProp

- Repeat until convergence.

#### b) Architectures of ANN

- Feedforward Neural Network (FNN): simple forward flow, no loops.

- Convolutional Neural Networks (CNN): specialized for images.

- Recurrent Neural Networks (RNN): handle sequential data (time series, text).

- Transformers: state-of-the-art for NLP and vision.

#### c) Regularization (Prevent Overfitting)

- Dropout: randomly deactivate neurons during training.

- L1/L2 Regularization: add penalty to large weights.

- Batch Normalization: stabilizes training by normalizing activations.

#### d) Practical Issues

- Vanishing/Exploding Gradients: solved using ReLU, better initialization, gradient clipping.

- Overfitting: solved with regularization, dropout, more data.

- Training Speed: solved using GPUs, parallelization, mini-batch training.

#### e) Advanced Improvements

- Deep Networks: more hidden layers = higher representational power.

- Residual Networks (ResNet): skip connections to solve vanishing gradients.

- Attention Mechanism: lets the model focus on important parts of data.

- Transfer Learning: reusing pretrained models on new tasks.
