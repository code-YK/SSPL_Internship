## Single neuron with Relu activation

```python
import numpy as np

# inputs
x = np.array([2, 3])     # two inputs
w = np.array([0.5, 0.7]) # weights
b = 1                    # bias

# weighted sum
z = np.dot(w, x) + b     

# activation (ReLU here)
y = np.maximum(0, z)     

print("Output:", y)
```

## Single Neuron with sigmod activation

```python
import numpy as np

# Inputs
x = np.array([1.0, 2.0, 3.0])      # 3 inputs
w = np.array([0.2, -0.5, 0.1])     # weights
b = -0.4                           # bias

# Weighted sum
z = np.dot(w, x) + b

# Sigmoid activation
y = 1 / (1 + np.exp(-z))

print("Output (manual sigmoid):", y)
```

## Using Dense Layer of 1 Neuron and Relu Activation

```python
import tensorflow as tf

# Define a single neuron (Dense layer)
neuron = tf.keras.layers.Dense(
    units=1,              # number of neurons
    activation='relu',    # activation function
    input_shape=(2,)      # number of inputs/features
)

# Create dummy input
x = tf.constant([[2.0, 3.0]])  # shape (1,2) -> batch of 1 with 2 features

# Pass input through neuron
output = neuron(x)
print("Output:", output.numpy())
```

## Using Dense Layer of 1 Neuron and Sigmod Activation

```python
import tensorflow as tf

# Define a single neuron with sigmoid
neuron = tf.keras.layers.Dense(
    units=1,
    activation='sigmoid',
    input_shape=(3,)
)

# Dummy input
x = tf.constant([[1.0, 2.0, 3.0]])
output = neuron(x)

print("Output (Dense sigmoid):", output.numpy())
```
