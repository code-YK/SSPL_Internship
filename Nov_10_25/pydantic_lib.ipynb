## Pydantic Library
- Pydantic is a Python library that uses Python type hints to:

    - Validate data at runtime

    - Enforce correct types

    - Automatically convert (coerce) data types

    - Provide useful error messages

- Pydantic brings runtime type enforcement and validation

### Basic Example

```python
from pydantic import BaseModel

class User(BaseModel):
    id: int
    name: str
    age: int
    is_admin: bool = False  # default value

# Valid data
user = User(id=1, name="Kuldeep", age=22)
print(user)

# Invalid data (auto-conversion will try first)
user2 = User(id="2", name="Ravi", age="23")
print(user2)
```

### Handling invalid Data type

```python
from pydantic import BaseModel, ValidationError

class User(BaseModel):
    id: int
    name: str
    age: int

try:
    user = User(id="abc", name="John", age="22")
except ValidationError as e:
    print(e)
```

### Optional Fields and Default Values

```python
from typing import Optional

class User(BaseModel):
    name: str
    email: Optional[str] = None
    age: int = 18

print(User(name="Alice"))
print(User(name="Bob", email="bob@example.com"))
```

### Nested Models
- embed one model inside another.

```python
class Address(BaseModel):
    city: str
    country: str

class User(BaseModel):
    name: str
    address: Address

user = User(name="Kuldeep", address={"city": "AhmedaBad", "country": "India"})
print(user)
print(user.address.city)
```

### Validation with Constraints
- Pydantic field constraints to enforce limits

```python
from pydantic import Field

class User(BaseModel):
    name: str = Field(min_length=3, max_length=50)
    age: int = Field(gt=0, lt=120)  # greater than 0, less than 120

user = User(name="Kuldeep", age=22)
print(user)


try:
    user_invalid = User(name="Al", age=150)
except ValidationError as e:
    print(e)
```

### Validators (Custom Logic)
- we can define our own validation logic using decorators.

```python
from pydantic import field_validator


class User(BaseModel):
    name: str
    age: int

    @field_validator("age")
    def check_age(cls, v):
        if v < 18:
            raise ValueError("User must be at least 18 years old")
        return v

user_1 = User(name="Ravi", age=20)  # valid
print(user_1)


try:
    user_2 = User(name="Sam", age=16)  # invalid
except ValidationError as e:
    print(e)
```

### Serialization and Parsing
- Convert model to dict or JSON

```python
user = User(name="Kuldeep", age=22)
print(user.model_dump())       # dict
print(user.model_dump_json())  # JSON
```

### API Like Data Validation

```python
from pydantic import BaseModel, EmailStr
from typing import List

class Student(BaseModel):
    name: str
    email: EmailStr
    scores: List[int]

data = {
    "name": "Kuldeep",
    "email": "kuldeep@example.com",
    "scores": [85, 90, 92]
}

student = Student(**data)
print(student.model_dump())
```

### Difference Between `typing` and `pydantic`

| Feature          | `typing`      | `pydantic`                 |
| ---------------- | ------------- | -------------------------- |
| Type enforcement | Only static | Runtime                  |
| Validation       | None        | Strong validation        |
| Auto conversion  | No          | Yes                      |
| Error messages   | None        | Detailed                 |
| Use case         | Type hinting  | Data models, APIs, configs |
