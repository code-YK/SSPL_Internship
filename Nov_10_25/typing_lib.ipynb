## Typing library

- The `typing` library in Python provides type hints - a way to specify the expected data types of variables, function arguments, and return values.

- It doesn’t enforce types at runtime, but it helps tools like mypy, Pyright, VSCode, and IDEs to catch bugs before running your code.

- Benefits:
    - Improves readability
    - Catches type-related bugs early
    - Helps IDEs with autocompletion and linting
    - Easier for teams to maintain and scale projects
    - Helpful for static analysis and documentation

### Basic Syntax :
    
    >def add(a: int, b: int) -> int:
    >    return a + b

- `a: int` → a must be an integer
- `b: int` → b must be an integer
- `-> int` → function returns an integer

```python
# typing variables
name: str = "Kuldeep"
age: int = 21
height: float = 5.9
is_student: bool = True
```

#### Union and Optional
- Union : A variable can be one of multiple types
- Optional : Means the value can also be None

```python
from typing import Union

data: Union[int, str] = "Hello"
data = 42  # valid
```

```python
from typing import Optional

name: Optional[str] = None
```

#### Callable (Functions as Arguments)

```python
from typing import Callable

def apply_function(x: int, func: Callable[[int], int]) -> int:
    return func(x)

def square(n: int) -> int:
    return n * n

def factorial(n: int) -> int:
    return 1 if n == 0 else n * factorial(n - 1)
    
print(apply_function(5, square))  # 25
print(apply_function(5, factorial))  # 120
```

#### Type Aliases

```python
from typing import List, Tuple

Vector = List[Tuple[int, int]]

points: Vector = [(1, 2), (3, 4)]
print(points)
```

#### TypedDict

```python
from typing import TypedDict

class User(TypedDict):
    id: int
    name: str
    email: str

user1: User = {"id": 1, "name": "Kuldeep", "email": "kul@example.com"}
print(user1)
```

#### Literal
- Restricts a variable to specific constant values

```python
from typing import Literal

def get_status(code: Literal["success", "error", "pending"]) -> str:
    return f"Status is {code}"

x = get_status("success")  # valid
print(x)

y = get_status("failed")   # type error
```

### Real Project example

```python
from typing import List, Dict, Optional

class Course:
    def __init__(self, title: str, instructor: str, students: Optional[List[str]] = None):
        self.title = title
        self.instructor = instructor
        self.students = students or []

    def enroll(self, student: str) -> None:
        self.students.append(student)

    def get_info(self) -> Dict[str, str]:
        return {"title": self.title, "instructor": self.instructor}

course = Course("Python", "Kuldeep")
course.enroll("Ravi")
print(course.get_info())
```
