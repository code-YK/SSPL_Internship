## Logging Library
- The logging library is Python’s built-in system for tracking events that happen during program execution.

#### Why to use logging :
- Because:
    - You can control what gets logged (info/debug/errors)

    - You can control where it goes (console, file, server, etc.)

    - You can format it neatly

    - You can disable it easily in production

#### Difference between Print and Logging
| `print()`                   | `logging`                               |
| --------------------------- | --------------------------------------- |
| Always outputs text         | Can be turned on/off                    |
| No levels or filtering      | Has levels (debug, info, warning, etc.) |
| Goes only to console        | Can go to file, syslog, email, etc.     |
| No timestamps or formatting | Full timestamps, context, etc.          |


```python
import logging

logging.basicConfig(level=logging.INFO)
logging.info(" --> Application started")
logging.warning(" --> Low disk space")
logging.error(" --> Something went wrong")
```

#### Logging Levels
- Each log message has a level of importance:

| Level      | Numeric Value | Usage                              |
| ---------- | ------------- | ---------------------------------- |
| `DEBUG`    | 10            | Detailed info for debugging        |
| `INFO`     | 20            | General information                |
| `WARNING`  | 30            | Something unexpected but not fatal |
| `ERROR`    | 40            | A serious problem                  |
| `CRITICAL` | 50            | Program may not continue           |


```python
logging.debug("Debug info")
logging.info("Normal info")
logging.warning("Potential issue")
logging.error("Error occurred")
logging.critical("Fatal error")
```

#### Logging to a File
- By default, logs go to console. But we can redirect them to a file:

```python
import logging

logging.basicConfig(
    filename="app.log",
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s", # Timestamp | Level | Message
    force=True  # Overwrite existing log file
)

logging.info("Started app")
logging.warning("This is a warning")
logging.error("This is an error")
```

```python
with open("app.log", "r") as f:
    print(f.read())
```

#### Creating Named Loggers
- Instead of using the root logger, create custom ones:

```python
import logging

logger = logging.getLogger("my_app")

logger.setLevel(logging.DEBUG)
logger.info("This is app-level logging")
```

#### Exception Logging

- You can log exceptions automatically:

```python
try:
    1 / 0
except ZeroDivisionError:
    logging.exception("An error occurred")
```

#### Cheat Sheet

| Concept           | Example                              | Purpose               |
| ----------------- | ------------------------------------ | --------------------- |
| Basic logging     | `logging.info("message")`            | Quick log             |
| Set level         | `basicConfig(level=logging.WARNING)` | Filter                |
| File logging      | `filename="app.log"`                 | Save to file          |
| Format            | `format="%(asctime)s ..."`           | Custom format         |
| Named logger      | `getLogger("my_app")`                | Module-specific logs  |
| Exception log     | `logging.exception("msg")`           | Capture traceback     |
| Multiple handlers | `StreamHandler()`, `FileHandler()`   | Console + File        |
| Rotation          | `RotatingFileHandler()`              | Auto-manage file size |


```python
import logging

# Configure once globally
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s | %(levelname)-8s | %(name)s | %(message)s",
    handlers=[
        logging.FileHandler("myapp.log"),
        logging.StreamHandler()
    ],
    force=True
)

logger = logging.getLogger("DemoApp")

logger.debug("Debug details")
logger.info("Starting application")
logger.warning("Low disk space")
logger.error("Something broke")
logger.critical("System failure!")
```
