## Single Agent powered by llm

- Goal: A single intelligent agent that can
    1. understand a task
    2. optionally use tools
    3. return a response

`User` → `Agent` → `(LLM + Instructions)` → `Response`

```python
from agno.agent import Agent
from agno.models.groq import Groq
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
api_key = os.getenv("GROQ_API_KEY")

# Create LLM
model = Groq(
    id="llama-3.1-8b-instant",
    api_key=api_key,
    temperature=0
)
```

```python
# Initialize Agent
agent = Agent(
    model=model,
    instructions="You are a helpful assistant")
```

```python
# Run Agent
response = agent.run("Explain agentic AI in simple words")
print(response.content)
```

```python
agent.print_response("What is the capital of France?")
```
