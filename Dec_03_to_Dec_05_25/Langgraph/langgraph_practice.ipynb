#### This is demo of to add llm in graphs. i am using gemini key.


`User → LLM Node → Output`

```python
from langgraph.graph import StateGraph, END
from typing import TypedDict, List
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, AIMessage
from IPython.display import display, Markdown
```

```python
# Load API key from .env file
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
# Create LLM
llm = ChatGoogleGenerativeAI(model="gemini-1.5-flash")
```

```python
# Define Agent State
class AgentState(TypedDict):
    messages: List[str]
```

```python
# Define LLM Node
def llm_node(state: AgentState):
    user_message = state["messages"][-1]
    response = llm.generate_content([user_message])
    state["messages"].append(response.text)
    return state
```

```python
# ---- GRAPH ----
graph = StateGraph(AgentState)
graph.add_node("llm", llm_node)
graph.set_entry_point("llm")
graph.add_edge("llm", END)

agent = graph.compile()
```

```python
# ---- RUN ----
state = {"messages": ["Explain quantum computing in simple words."]}

result = agent.invoke(state)
print(result["messages"][-1])
```
