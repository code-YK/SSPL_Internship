```python
import sqlite3
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy.orm import sessionmaker
```

```python
db_path = 'sqlite:///demo.db'

engine = create_engine(db_path)
Base = declarative_base()
Session = sessionmaker(bind=engine)
session = Session()
```

```python
class Superhero(Base):
    __tablename__ = 'superheroes'
    id = Column(Integer, primary_key=True)
    name = Column(String)
    power = Column(String)
    strength = Column(Integer)

Base.metadata.create_all(engine)
```

```python
hero1 = Superhero(name='Superman', power='Flight', strength=100)
hero2 = Superhero(name='Batman', power='Martial Arts', strength=85)
hero3 = Superhero(name='Flash', power='Super Speed', strength=75)
hero4 = Superhero(name='Wonder Woman', power='Super Strength', strength=95)
hero5 = Superhero(name='Aquaman', power='Underwater Breathing', strength=80)
hero6 = Superhero(name='Cyborg', power='Technopathy', strength=90)

session.add_all([hero1, hero2, hero3, hero4, hero5, hero6])
session.commit()
```

```python
# select and display all records
all_heroes = session.query(Superhero).all()

import tabulate
table = [(hero.id, hero.name, hero.power, hero.strength) for hero in all_heroes]
print(tabulate.tabulate(table, headers=["ID", "Name", "Power", "Strength"], tablefmt="psql"))
```

```python
# Adding a new superhero
hero7 = Superhero(name='Iron Man', power='Powered Armor Suit',strength=88)
hero8 = Superhero(name='Thor', power='God of Thunder', strength=95)
hero9 = Superhero(name='Hulk', power='Super Strength', strength=100)
hero10 = Superhero(name='Black Widow', power='Espionage', strength=75)
hero11 = Superhero(name='Hawkeye', power='Archery', strength=80)
hero12 = Superhero(name='Doctor Strange', power='Magic', strength=90)
hero13 = Superhero(name='Captain America', power='Super Soldier Serum', strength=95)
hero14 = Superhero(name='Spider-Man', power='Spider Abilities', strength=85)

session.add_all([hero7, hero8, hero9, hero10, hero11, hero12, hero13, hero14])
session.commit()
```

```python
# Select heroes with strength greater than 90
strong_heroes = session.query(Superhero).filter(Superhero.strength > 90).all()
strong_table = [(hero.id, hero.name, hero.power, hero.strength) for hero in strong_heroes]
print("\nHeroes with Strength > 90:")
print(tabulate.tabulate(strong_table, headers=["ID", "Name", "Power", "Strength"], tablefmt="psql"))
```

```python
# select heroes whose names start with 'S'
s_heroes = session.query(Superhero).filter(Superhero.name.like('S%')).all()
s_table = [(hero.id, hero.name, hero.power, hero.strength) for hero in s_heroes]
print("\nHeroes whose names start with 'S':")
print(tabulate.tabulate(s_table, headers=["ID", "Name", "Power", "Strength"], tablefmt="psql"))
```

```python
from sqlalchemy import and_, or_, not_

# Select heroes with strength between 80 and 90
between_heroes = session.query(Superhero).filter(
    and_(Superhero.strength >= 80, Superhero.strength <= 90)
    ).all()
between_table = [(hero.id, hero.name, hero.power, hero.strength) for hero in between_heroes]
print("\nHeroes with Strength between 80 and 90:")
print(tabulate.tabulate(between_table, headers=["ID", "Name", "Power", "Strength"], tablefmt="psql"))
```

#### This database models a platform where users enroll in courses, instructors create courses, and students leave reviews.

```python
db_path = 'sqlite:///education_platform.db'

engine = create_engine(db_path)
Base = declarative_base()
Session = sessionmaker(bind=engine)
session = Session()
```

```python
from sqlalchemy import ForeignKey, Column, Integer, String, Text, Date, Boolean
from sqlalchemy.orm import relationship
from datetime import date
import tabulate
```

- tables

```python
class User(Base):
    __tablename__ = 'users'

    user_id = Column(Integer, primary_key=True)
    name = Column(String)
    email = Column(String, unique=True)
    country = Column(String)
    is_instructor = Column(Boolean, default=False)

class Course(Base):
    __tablename__ = 'courses'

    course_id = Column(Integer, primary_key=True)
    title = Column(String)
    category = Column(String)
    instructor_id = Column(Integer, ForeignKey("users.user_id"))
    price = Column(Integer)
    created_at = Column(Date, default=date.today)

    # Relationship to User
    instructor = relationship("User", backref="courses")

class Enrollment(Base):
    __tablename__ = 'enrollments'

    enrollment_id = Column(Integer, primary_key=True)
    user_id = Column(Integer, ForeignKey("users.user_id"))
    course_id = Column(Integer, ForeignKey("courses.course_id"))
    enrolled_at = Column(Date, default=date.today)

    # Relationships
    user = relationship("User", backref="enrollments")
    course = relationship("Course", backref="enrollments")

class Review(Base):
    __tablename__ = 'reviews'

    review_id = Column(Integer, primary_key=True)
    user_id = Column(Integer, ForeignKey("users.user_id"))
    course_id = Column(Integer, ForeignKey("courses.course_id"))
    rating = Column(Integer)
    comment = Column(Text)

    # Relationships
    user = relationship("User", backref="reviews")
    course = relationship("Course", backref="reviews")
```

```python
Base.metadata.create_all(engine)
```

### Q1 => Get all courses priced under ₹1500.

```python
q1 = session.query(Course).filter(Course.price < 1500).all()
q1_table = [(course.course_id, course.title, course.category, course.price) for course in q1]
print("Courses priced under ₹1500:")
print(tabulate.tabulate(q1_table, headers=["Course ID", "Title", "Category", "Price"], tablefmt="psql"))
```

### Q2 => Show all users from India who are not instructors.

```python
q2 = session.query(User).filter(
    and_(User.country == "India", User.is_instructor == False)
).all()

q2_table = [(user.user_id, user.name, user.email, user.country) for user in q2]
print("Users from India who are not instructors:")
print(tabulate.tabulate(q2_table, headers=["User ID", "Name", "Email", "Country"], tablefmt="psql"))
```

### Q3 => Show course title, instructor name, and price.

```python
q3 = session.query(Course.title, User.name, Course.price).join(
    User, Course.instructor_id == User.user_id
).all()

q3_table = [(title, name, price) for title, name, price in q3]
print("Courses and their instructors:")
print(tabulate.tabulate(q3_table, headers=["Course Title", "Instructor Name", "Price"], tablefmt="psql"))
```

### Q4 => Show the courses that got at least one 5-star review.

```python
q4 = session.query(Course.title, Course.category, Review.rating, Review.comment).join(
    Review, Course.course_id == Review.course_id
    ).filter(Review.rating == 5).all()
q4_table = [(title, category, rating, comment) for title, category, rating, comment in q4]
print("Courses that got at least one 5-star review:")
print(tabulate.tabulate(q4_table, headers=["Title", "Category", "Rating", "Comment"], tablefmt="psql"))
```

#### Q7 => List all users who haven't left any reviews.

```python
q7 = session.query(User).outerjoin(Review).filter(Review.review_id == None).all()
q7_table = [(user.user_id, user.name, user.email) for user in q7]
print("Users who haven't left any reviews:")
print(tabulate.tabulate(q7_table, headers=["User ID", "Name", "Email"], tablefmt="psql"))
```

#### Q8 => Get the category with the most active enrollments.

```python
q8 = session.query(Course.category, func.count(Enrollment.enrollment_id).label('enrollment_count')).join(
    Enrollment, Course.course_id == Enrollment.course_id
).group_by(Course.category).order_by(func.count(Enrollment.enrollment_id).desc()).first()
print(f"Category with the most active enrollments: {q8.category} ({q8.enrollment_count} enrollments)")
```

#### Q9 => Rank courses by number of enrollments.

```python
q9 = session.query(Course.title, func.count(Enrollment.enrollment_id).label('enrollment_count')).join(
    Enrollment, Course.course_id == Enrollment.course_id
).group_by(Course.title).order_by(func.count(Enrollment.enrollment_id).desc()).all()
q9_table = [(title, count) for title, count in q9]
print("Courses with the most enrollments:")
print(tabulate.tabulate(q9_table, headers=["Course Title", "Enrollment Count"], tablefmt="psql"))
```

#### Q10 => Show average price per category along with each course.

```python
q10 = session.query(Course.title, Course.category, Course.price, func.avg(Course.price).over(partition_by=Course.category).label('avg_price_per_category')).all()
q10_table = [(title, category, price, round(avg_price, 2)) for title, category, price, avg_price in q10]
print("Courses with average price per category:")
print(tabulate.tabulate(q10_table, headers=["Course Title", "Category", "Price", "Avg Price per Category"], tablefmt="psql"))
```
