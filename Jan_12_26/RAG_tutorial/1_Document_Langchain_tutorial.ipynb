## Document Loader

#### A Document Loader in LangChain is used to load raw data from a source (file, website, PDF, API, database, etc.) and convert it into a standard format called a Document.

#### A Document contains:

- page_content → the actual text
- metadata → extra information (source, page number, URL, etc.)

#### Why it’s needed ?

- LLMs can’t read files directly
- Loaders standardize data for:
    - Text splitting
    - Embeddings
    - Vector stores
    - RAG pipelines

LangChain internally uses this structure:

```python 
Document(
  page_content="actual text",
  metadata={"source": "..."}
)
```

````python
from langchain_core.documents import Document

doc = Document(
    page_content="LangChain makes LLM applications easier.",
    metadata={"source": "manual"}
)

print(doc.page_content)
print(doc.metadata)
````

## DIRECTORY LOADER

- Loads all files inside a directory
- Useful when we have:
    - Logs
    - Notes
    - Articles
    - Datasets

````python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

loader = DirectoryLoader(
    path="data/", # directory containing text files
    glob="**/*.txt", # pattern to match text files recursively
    loader_cls=TextLoader # specify the loader class to use (TextLoader for .txt files)
)

documents = loader.load()
print(len(documents))

print('==='*20)
for doc in documents:
    print(doc.page_content)
    print(doc.metadata)
    print('---'*20)
print('==='*20)
````

## PDF LOADER
- Reads PDF files page by page

- Each page becomes one Document

- `PyPDFLoader`, `PyMuPDFLoader` 

````python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("E:\\SSPL_Internship_Repo\\Jan_12_25\\RAG_tutorial\\data\\demo.pdf")
documents = loader.load()

print(documents[0].page_content)
print(documents[0].metadata)
````

## WEBBASE LOADER

## CSV LOADER

- Loads CSV rows as documents
- Each row = one document

````python
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(
    file_path="E:\\SSPL_Internship_Repo\\Jan_12_25\\RAG_tutorial\\data\\data.csv",
    source_column="Project Name"
)

documents = loader.load()
documents[0]
````

## JSON LOADER

- Loads structured JSON data
- We specify:
    - JSON path
    - What field to extract

````python
from langchain_community.document_loaders import JSONLoader

loader = JSONLoader(
    file_path="E:\\SSPL_Internship_Repo\\Jan_12_25\\RAG_tutorial\\data\\data.json",
    jq_schema=".employees[]",
    text_content=False
)

documents = loader.load()
documents[0]
````

## UNSTRUCTURED LOADER

- Unstructured loaders work for:
    - PDFs
    - Word docs
    - PPTs
    - Emails

- They auto-detect structure.

````python
from langchain_community.document_loaders import UnstructuredWordDocumentLoader

loader = UnstructuredWordDocumentLoader("E:\\SSPL_Internship_Repo\\Jan_12_25\\RAG_tutorial\\data\\demo.docx")
documents = loader.load()

print(documents[0].page_content)
print(documents[0].metadata)
````
