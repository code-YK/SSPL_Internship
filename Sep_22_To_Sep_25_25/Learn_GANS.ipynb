## Types of GANs

### 1. **Vanilla GAN**
The original GAN by Goodfellow (2014).  
- **Generator** and **Discriminator** are just simple feedforward neural networks.  
- Good for learning the concept, but struggles with complex data like high-res images.  

---

### 2. **DCGAN (Deep Convolutional GAN)**
Uses **convolutional layers** instead of simple dense layers.  
- Much better at generating realistic images.  
- Stable training compared to Vanilla GANs.  
- Great for image data (faces, objects).  

---

### 3. **Conditional GAN (cGAN)**
Adds **labels** as input (e.g., "generate a cat" vs. "generate a dog").  
- Used when you want **control** over what is generated.  
- Example: Generate images of shoes in a specific style.  

---

### 4. **CycleGAN**
Used for **image-to-image translation** without paired data.  
- Example: Turn a photo of a horse into a zebra, or a summer scene into winter — even if you don’t have exact before/after image pairs.  

---

### 5. **StyleGAN**
Super advanced GAN (used by *ThisPersonDoesNotExist.com*).  
- Can generate **very realistic, high-resolution faces**.  
- Gives control over **style** (age, smile, hair, etc.) of generated images.  

---

### 6. **Pix2Pix**
Similar to CycleGAN, but needs **paired data**.  
- Example: You have a black-and-white image and its colored version → train Pix2Pix to colorize new B/W images.  

---

### 7. **Wasserstein GAN (WGAN)**
Solves the problem of unstable training by using a different loss function (Wasserstein distance).  
- More stable.  
- Better quality outputs.

# GAN Workflow – Step by Step

## 1️⃣ Imports & Setup
- Import PyTorch libraries: `torch`, `torch.nn`, `torch.optim`.
- Decide device: `device = torch.device("cuda" if torch.cuda.is_available() else "cpu")`.

---

## 2️⃣ Noise / Latent Vector
- Generator starts from random noise `z` (e.g., size 100).  
- `z` is transformed by the Generator into fake images.

---

## 3️⃣ Define the Generator
- Input: `z` (shape `[batch, 100]`)  
- Layers: `Linear(100→256) → ReLU → Linear(256→784) → Tanh`  
- Output: `[batch, 784]` → reshape to `28×28`.  
- `Tanh` outputs in `[-1,1]` (match normalized data).

---

## 4️⃣ Define the Discriminator
- Input: flattened image `[784]`  
- Layers: `Linear(784→256) → LeakyReLU → Linear(256→1) → Sigmoid`  
- Output: probability (1 = real, 0 = fake)

---

## 5️⃣ Instantiate Models
- `G = Generator().to(device)`  
- `D = Discriminator().to(device)`

---

## 6️⃣ Loss Function
- Binary Cross-Entropy: `loss_fn = nn.BCELoss()`  
- Compares predicted probability vs true label (1 for real, 0 for fake)

---

## 7️⃣ Optimizers
- `optimizer_G = Adam(G.parameters(), lr=0.0002)`  
- `optimizer_D = Adam(D.parameters(), lr=0.0002)`

---

## 8️⃣ Training Loop (Epochs)
- Repeat for each epoch and each batch in dataset

---

## 9️⃣ Train Discriminator (D)
1. Zero gradients: `optimizer_D.zero_grad()`
2. **Real pass:** flatten real images, label = 1, compute loss
3. **Fake pass:** generate fake images, label = 0, compute loss (`fake_data.detach()` to not update G)
4. Total loss: `loss_D = loss_real + loss_fake`
5. Backprop and update D: `loss_D.backward()`, `optimizer_D.step()`

---

## 🔟 Train Generator (G)
1. Zero gradients: `optimizer_G.zero_grad()`
2. Generate fake images
3. Compute loss vs **real labels** (G wants to fool D)
4. Backprop and update G: `loss_G.backward()`, `optimizer_G.step()`

---

## 1️⃣1️⃣ Monitor & Output
- Print `loss_D` and `loss_G` per epoch
- Periodically generate and visualize fake images
- Save model checkpoints

---

## Tips & Notes
- Normalize inputs to match Tanh (`[-1,1]`)
- Use BatchNorm in G and LeakyReLU in D for stability
- Watch for mode collapse (G produces same outputs)
- Adjust learning rates and monitor gradients
