## OpenCV 

OpenCV (Open Source Computer Vision Library) is a computer vision and image processing library written in C++ with Python bindings.

It helps you do things like:

- Detect objects, faces, motion.
- Work with images and videos.
- Use pre-trained deep learning models (like YOLO, SSD, etc.).
- Perform tasks like OCR, segmentation, or edge detection.

## 1. Reading and Displaying image

- `cv2.imread()` → loads the image.

- `cv2.imshow()` → opens a window to display it.

- `cv2.waitKey(0)` → waits indefinitely for a key press.

- `cv2.destroyAllWindows()` → closes the window.

```python
import cv2
from matplotlib import pyplot as plt

# Load the image
img = cv2.imread(r"E:\SSPL_Internship_Repo\Oct_07_25\openCV_learnings\data\building.jpg")

# Check if image loaded correctly
if img is None:
    print("Image not loaded! Check path.")
else:
    # Convert BGR to RGB for correct colors
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Display using Matplotlib
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.title('Original Image')
    plt.show()
```

```python
# Function to display image using matplotlib
def show_image(image, title='Image'):
    plt.imshow(image)
    plt.axis('off')
    plt.title(title)
    plt.show()
```

### 2. Resize Image

```python
resized_img = cv2.resize(img_rgb, (200, 300))  # Width x Height
show_image(resized_img, 'Resized Image')
```

### 3. Crop Image

```python
# Crop Image (y1:y2, x1:x2)
cropped_img = img_rgb[100:250, 100:400]  # Adjust as needed
show_image(cropped_img, 'Cropped Image')
```

### 4. Grayscale

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
show_image(gray_img, 'Grayscale Image')
```

### 5.Save images or output

```python
# Save Images
cv2.imwrite('resized_image.jpg', cv2.cvtColor(resized_img, cv2.COLOR_RGB2BGR))
cv2.imwrite('cropped_image.jpg', cv2.cvtColor(cropped_img, cv2.COLOR_RGB2BGR))
cv2.imwrite('grayscale_image.jpg', gray_img)
```
